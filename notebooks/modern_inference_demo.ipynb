# Modern Inference Demo

Este notebook resume el entrenamiento final del ganador y enseña directamente las salidas visuales de inferencia sobre las imágenes de webcam.

## Qué se muestra aquí

- resumen del modelo ganador
- métricas finales bloqueadas sobre la seed 42
- tabla de predicciones con score y etiqueta
- imágenes exportadas por el visualizador de `anomalib`

```python
from pathlib import Path
import json

import pandas as pd
from IPython.display import Image, display

PROJECT_ROOT = None
for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
    if (candidate / 'artifacts').exists() and (candidate / 'README.md').exists():
        PROJECT_ROOT = candidate
        break
if PROJECT_ROOT is None:
    raise RuntimeError('Could not locate the project root from the current working directory.')

final_root = PROJECT_ROOT / 'artifacts' / 'modern' / 'final_model'
predictions_root = PROJECT_ROOT / 'artifacts' / 'modern' / 'predictions'

summary = json.loads((final_root / 'final_model_summary.json').read_text(encoding='utf-8'))
predictions = pd.read_csv(predictions_root / 'folder_predictions.csv')

summary
```

Aquí ya aparecen el score y la decisión binaria del modelo para cada imagen.

## Visualizaciones de inferencia

Estas PNG las genera directamente `anomalib` y suelen incluir la imagen original, el mapa de anomalía y la salida final compuesta.

```python
for _, row in predictions.iterrows():
    print(f"{Path(row['image_path']).name} | score={row['pred_score']:.4f} | pred_label={row['pred_label']}")
    if isinstance(row.get('visualization_path'), str) and Path(row['visualization_path']).exists():
        display(Image(filename=row['visualization_path'], width=720))
    else:
        print('No visualization file found for this image.')
```
